- **Main research question:** To what extent can different dimensions of problematic mobile-phone use predict smartphone addiction severity among students?

- **Who might use or care about the answer?** Researchers, schools, parents, psychologists, and organisations focused on youth wellbeing.
- **Possible sub-questions:** 
Which phone-use behaviours are most linked to addiction?
Does withdrawal predict addiction severity?
Does compulsive use predict addiction severity?

### 1. Where did the data come from?

Answer: The data came from a research study conducted in China, with participants recruited from different educational levels. The final dataset contains 1,619 participants: 738 middle-school students, 654 undergraduate students and 227 graduate students. There were 628 males and 991 females.


### 2. Why was it originally collected?

**Answer:** The data was originally collected to study problematic mobile-phone use and develop a better way to measure smartphone addiction using different questionnaires and addiction-related behaviours.

### 3. Who is represented?

**Answer:** 
 The dataset contains 1,619 participants: 738 middle-school students, 654 undergraduate students and 227 graduate students. There were 628 males and 991 females.


### 4. Who might be missing?

**Answer:** People outside the participating schools and universities, people from other countries, and people from different age groups may be missing. The dataset may also underrepresent people who do not regularly use smartphones.

### 5. What personal or sensitive information does it contain, explicitly or implicitly?

**Answer:** It contains personal information such as gender and education level, as well as sensitive information about smartphone addiction, withdrawal, compulsive use, mood changes, and social difficulties. The data is anonymous, so participants' names and identifying details are not included.

### 6. What could your analysis reveal that was not obvious in the original data?

**Answer:** Our analysis could reveal which behaviours are the strongest predictors of smartphone addiction severity and whether we can accurately predict addiction levels using machine learning.

### 7. Who could benefit from your findings?

**Answer:** Known evidence: Students, parents, teachers, schools, and youth organisations could benefit from the findings by learning more about problematic smartphone use.
Reasonable interpretation: The results could help them understand which behaviours are most strongly linked to smartphone addiction and identify areas that may need more attention.
